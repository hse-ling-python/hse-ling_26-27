## Основы ООП: создание класса, определение атрибутов и методов

**Объектно-ориентированное программирование** (ООП) — один из ключевых подходов к программированию. В нем программы представляются как набор **объектов**, каждый из которых — это представитель некоторого более общего типа (**класса**).

В питоне все — объект. Списки (`list`), множества (`set`), словари (`dict`) — это объекты.

Каждый объект обладает **свойствами** (или атрибутами) и **действиями** (или методами), которые с этими объектами можно выполнять.

Например, в питоне мы умеем складывать не только числа и строки, но и списки. Мы привыкли просто пользоваться этими структурами данных, но никогда особо не задумывались – как они устроены изнутри. ООП помогает в этом разобраться.

**Зачем нам ООП?**

Помимо встроенных структур данных нам может быть удобно научиться создавать свои собственные **классы**.

Пример: вы хотите создать свой собственный словарь (например, русско-английский). С одной стороны, вы хотите хранить много мета-информации о самом словаре. С другой стороны, вы хотите быстро получать переводы слов, добавлять новые пары, возможно, какие-то глоссы и т.д.

Можно придумать структуру на основе обычного встроенного словаря (`dict`), а можно написать **свой собственный класс**.


**Функционал нашего будущего русско-английского словаря:**

*   атрибуты / свойства
    *   исходный язык
    *   язык перевода
    *   количество слов в словаре
    *   авторы словаря
    *   дата создания словаря
*   методы / действия
    *   получить перевод слова с русского на английский язык
    *   добавить новое слово
    *   удалить слово
    *   поменять/уточнить перевод
    *   получить список всех слов словаря



**Как создаются класс, атрибуты и методы?**

*   По сути методы — это функции внутри класса (обычные с `def`),  а атрибуты — это переменные внутри него (но не все).
*   Классы обычно называются с большой буквы, а объекты-представители с маленькой. Для создания класса нужно ключевое слово `class`.
*   У каждого класса должен быть метод `__init__` — это **конструктор**, он отвечает за создание (или инициализацию) объекта (экземпляра класса). Этот метод может принимать дополнительные аргументы, необходимые для создания объекта (это уже решаем мы как разработчики). А может не принимать. Но **всегда** первым аргументом идет `self` – это экземпляр класса, он отвечает за работу с самим классом изнутри.

Игрушечный пример для иллюстрации основного синтаксиса:

```
class SomeClass:
    def __init__(self):
        self.some_attribute = 0
    
    def some_method(self):
        return self.some_attribute
```

Теперь разберем код, который создает класс русско-английского словаря:

````python
from datetime import datetime

class Dictionary:
    # инициализируем класс и определяем атрибуты
    def __init__(self, source=None, target=None):
        self.source_language = source
        self.target_language = target
        self.date = datetime.now()
        self.vocabulary = {} # словарь dict для хранения данных внутри нашего класса

    # добавление нового слова в словарь
    def add_word(self, source, target):
        if source not in self.vocabulary:
            self.vocabulary[source] = target
        else:
            print(f'{source} is already added')

    # удаление слова из словаря
    def delete_word(self, word):
        if word in self.vocabulary:
            self.vocabulary.pop(word)
        else:
            print(f'{word} not in dictionary')

    # получаем все слова из словаря
    def get_all_words(self):
        return self.vocabulary
````

Создадим словарь и добавим туда несколько слов, чтобы проверить, что все работает!

````python
def test(my_dict, src='кошка', tgt='cat'):
    print(f'Сам словарь: {my_dict}', end='\n\n')

    print('Атрибуты')
    print(f'Исходный язык: {my_dict.source_language}')
    print(f'Язык перевода: {my_dict.target_language}')
    print(f'Дата создания: {my_dict.date}', end='\n\n')

    print('Методы')
    print(f'Добавление слова:')
    my_dict.add_word(source=src, target=tgt)
    print(f'Все слова в словаре: {my_dict.get_all_words()}')
    print('Удаление слова:')
    my_dict.delete_word(word=src)
    print(f'Все слова в словаре: {my_dict.get_all_words()}')

dct = Dictionary('ru', 'en')
test(dct)
````

## Три ~~четыре~~ кита ООП

1. **Инкапсуляция**: мы храним вместе объект и связанные с ним характеристики и действия (атрибуты и методы).
2. **Наследование**: мы можем перенимать атрибуты и методы у другого класса (наследовать их).
3. **Полиморфизм**: мы можем использовать одинаковый интерфейс у разных объектов.

И четвертый не совсем кит — **абстракция**. Ее суть заключается в том, что пользователь должен иметь доступ только к интерфейсу, но не к деталям внутренней реализации класса.

### Абстракция

Когда мы писали класс словаря, мы подразумевали, что пользователь будет что-то добавлять или удалять с помощью специальных методов. Вот только пользователям чаще всего **не очень** интересно, что мы подразумевали – они могут вести себя как захотят.

Здесь возникает **абстракция**: мы хотим, чтобы снаружи было доступно только то, что изначально было на это рассчитано.

````python
# вот так
dct.add_word('мяч', 'ball')

# а не вот так
dct.vocabulary['питон'] = 'python'

dct.get_all_words()
````

К сожалению, у питона конкретно с этим принципом все не очень хорошо: как бы вы что ни «прятали», пользователь все равно все сможет достать.

Но как спрятать что-то от пользователя?
Можно поставить один или два знака `_` в начале названия атрибута или метода — тогда питон может выдасть ошибку, если пользователь будет обращаться к таким атрибутам или методам напрямую.

В питоне приняты три вида названий атрибутов и методов:


*   Просто названия с маленькой буквы (`public`): доступны снаружи и формируют интерфейс для пользователя
*   Названия, которые начинаются с `_` (одно подчеркивание, `protected`, но не совсем): доступны пользователю, но мы мягко намекаем, что это можно трогать только, если он очень уверен в своих действиях
*   Названия с `__` (два подчеркивания, private): недоступны снаружи, но их все еще можно вытащить методом хитрых комбинаций (пользователь точно знает, что лезет куда-то не туда)

На мини-версии нашего класса посмотрим, как это работает:

````python
from datetime import datetime

class SafeDictionary:
    def __init__(self, source=None, target=None):
        self.source_language = source
        self.target_language = target
        self.date = datetime.now()

        # спрячем этот атрибут
        self.__vocabulary = {}

    def add_word(self, source, target):
        if source not in self.__vocabulary:
            self.__vocabulary[source] = target
        else:
            print(f'{source} is already added')

    def delete_word(self, word):
        if word in self.__vocabulary:
            self.__vocabulary.pop(word)
        else:
            print(f'{word} not in dictionary')

    def get_all_words(self):
        return self.__vocabulary
````

Теперь питон на нас ругается при попытке достать словарь напрямую, нам придется пользоваться методом, который мы сами и написали – `get_all_words`.

### Инкапсуляция

Класс, который мы написали – это пример **инкапсуляции**. Мы в **одном классе** храним и атрибуты, и методы.

Как делать **НЕ** надо:

````python
class SillyDictionary:
    def __init__(self, source=None, target=None):
        self.source_language = source
        self.target_language = target
        self.date = datetime.now()
        self.vocabulary = {}

    def add_word(self, source, target):
        if source not in self.vocabulary:
            self.vocabulary[source] = target
        else:
            print(f'{source} is already added')

    def delete_word(self, word):
        if word in self.vocabulary:
            self.vocabulary.pop(word)
        else:
            print(f'{word} not in dictionary')

def get_all_words(dct):
    return dct.vocabulary

dct = SillyDictionary('ru', 'en')
dct.add_word('мяч', 'ball')
get_all_words(dct)
````

### Наследование

Сейчас у нас есть словарь, с которым все просто: слова всегда подаются по одному. Но вдруг нам понадобилось добавлять слова списками. Здесь нам может помочь **наследование**.

Идея: мы не хотим переписывать общие для старого и нового словаря методы (например, заново прописывать инициализацию атрибутов или вывод всех слов словаря). Однако мы хотим создать **новый метод**, который умеет добавлять список новых слов.

Для этого можно создать новый класс, но унаследовать его от предыдущего.

````python
# класс, от которого мы наследуемся указывается
# в круглых скобках при создании нового класса
class DictionaryWithList(SafeDictionary):
    # добавляем слова из списка words
    def add_many_words(self, words):
        for word, target in words:
            self.add_word(word, target)

    # удаляем слова из списка words
    def delete_many_words(self, words):
        # удаляем много слов
        for word in words:
            self.delete_word(word)
````

````python
dct = DictionaryWithList('ru', 'en')
````

````python
# тут вызывается метод родительского класса
dct.add_word('кошка', 'cat')
````

````python
words = [('собака', 'dog'), ('птица', 'bird'), ('стол', 'table')]
# а тут – новый метод, который мы определили
dct.add_many_words(words)
dct.get_all_words()
````

### Полиморфизм

Идея **полиморфизма** в том, что под одинаковым интерфейсом (название метода или оператора) может скрываться разное содержание, реализующее нужный нам метод. При этом то, что переопределять не надо, мы никак не меняем.

Например, можно снова вспомнить `+` для строк, чисел или списков. Оператор один – а функционал разный.

Мы можем добавить в словарь следующий функционал: перед тем как добавить что-то в словарь, мы лемматизируем входные данные. При этом нам бы очень не хотелось объяснять пользователям, что теперь им нужно использовать метод с другим именем – поэтому будем переопределять только метод `add_word`.

````python
import pymorphy3

morph = pymorphy3.MorphAnalyzer()
````

````python
class SuperDictionary(SafeDictionary):
    def add_word(self, source, target):
        # получаем лемму
        word = morph.parse(source)[0].normal_form
        # добавляем слово – метод родителя!
        SafeDictionary.add_word(self, word, target)
````

````python
dct = SuperDictionary('ru', 'en')

words = [('собаки', 'dog'), ('птица', 'bird'), ('столы', 'table')]
for source, target in words:
    dct.add_word(source, target)

dct.get_all_words()
````

### Принципы **SOLID**

**S — Single Responsibility Principle** — у класса должна быть только одна четкая цель и логика использования. Швейцарские ножи в мире ООП не поощряются.

**O — Open Closed Principle** — мы должны иметь возможность расширять возможности класса (добавить новый метод, переопределить старый), но у нас не должно возникать необходимости править исходный код.

**L — Liskov Substitution Principle** — наследник может использоваться на месте родителя (словарь с красивым интерфейсом - все еще словарь и в нем должно быть можно найти нужное слово).

**I — Interface Segregation Principle** — класс должен уметь только то, что уместно для его зоны отвественности (русско-английский словарь не умеет рисовать картины и переводить книги, он просто словарь).

**D — Dependency Inversion Principle** — не должно быть прямых зависимостей между модулями, только между интерфейсами (у переводчика не должно быть методов «перевести на русский», «перевести на английский», только метод «перевести», который использует интерфейс словарей).

Более подробно можно почитать вот [тут](https://habr.com/ru/articles/688530/).


### `*args` и `**kwargs`

Они позволяют очень удобно собрать все аргументы функции в один кортеж и(ли) словарь – позиционные уйдут в `*args` и кортеж, а именованные уйдут в `**kwargs` и словарь.

````python
def func(*args, **kwargs):
    print(f'args: {args}')
    print(f'kwargs: {kwargs}')

func(1, 2, 3, 5, a=23, b=12, c=74)
````

### `@staticmethod`

**Декоратор** — это функция, которая принимает другую функцию и возвращает новую, расширенную версию. Синтаксис `@decorator` — это просто сокращение для `func = decorator(func)`.

В питоне есть встроенные декораторы:
`@staticmethod`, `@classmethod` и `@property`. Они работают по тому же принципу: принимают метод и возвращают новую версию с изменённым поведением.

Иногда метод объекта никак не использует его атрибуты (не обращается к `self` и **не принимает его в качестве аргумента**), а просто выполняет независимую логику. В таком случае такой метод называется статическим.

Мы можем использовать на нём специальный декоратор `@staticmethod`, и тогда мы помимо прочего сможем вызывать его, даже не создавая объект.

````python
class StrangeDictionary(Dictionary):
    @staticmethod
    def strange_method(word):
        print(word, len(word))

    # если не указать, что метод static
    def very_strange_method(word):
        print(word, len(word))

dct = StrangeDictionary()
dct.strange_method('кошка')
````

````python
dct.very_strange_method('кошка')
````

### Перегрузка операторов

Для классов мы можем переопределять базовые операции – здесь опять можно вспомнить пример со сложением разных чисел, строк и списков, для которых используется один и тот же оператор `+`.

Например, `__repr__` позволяет рассказать питону, как надо отображать объект этого класса, а `__str__` — как надо его переводить в строку.

````python
class BeautifulDictionary(SuperDictionary):
    def __repr__(self):
        s = type(self).__name__.split('.')[-1]
        args = []
        for el, val in self.__dict__.items():
            args.append(f'{el}={val}')
        if len(args) > 0:
            args = f'({", ".join(args)})'
            s += args
        return s

dct = BeautifulDictionary('ru', 'en')

words = [('собаки', 'dog'), ('птица', 'bird'), ('столы', 'table')]
for source, target in words:
    dct.add_word(source, target)

print(dct)
````

### Наследование: `super()` и MRO

Если при наследовании в классе-ребенке мы хотим переопределить методы класса-родителя, то нам понадобится оператор `super()`.

Все методы, вызывающиеся от `super()`, будут вызваны от родительского класса (классов) объекта. Таким образом, когда мы вызываем `super().__init__(name)` в классах-детях мы выполняем код из метода `__init__` родительского класса.

````python
class SuperDictionary(SafeDictionary):
    # добавим в инициализацию новый атрибут – название используемого анализатора
    def __init__(self, source=None, target=None):
        super().__init__(source, target)
        self.analyzer = 'pymorphy'

    def add_word(self, source, target):
        # получаем лемму
        word = morph.parse(source)[0].normal_form
        # добавляем слово – метод родителя!
        SafeDictionary.add_word(self, word, target)

dct = SuperDictionary('ru', 'en')
print(dct.analyzer)
````

Один класс может быть ребёнком сразу нескольких родителей-классов. В таком случае они перечисляются через запятую – `class Children(A, B)`.

Что делать, если у `A` и `B` есть метод с одинаковым именем? Питон вызовет тот, что нашёл первым по MRO.
