## Зачем нам это все?

**Пример №1**

Можно же просто установить нужное. Однако после установки код все равно не запустился, так как у вас оказалась другая версия питона, соответственно, другие версии библиотек – и функции в них называются по-другому.

Чтобы такой проблемы избежать, нам необходимо использовать **файл зависимостей** – файл в специальном формате, где прописаны не только названия библиотек, но и их версии (а часто – уникальные хэши конкретных файлов библиотеки, чтобы устанавливалось __точно__ то, что нужно).

Однако, если у вас несколько проектов, то установка новых зависимостей под каждый может привести к конфликту. Чтобы этого избежать, нам необходимо использовать **виртуальное окружение** - изолированную пару из интерпретатора питона и установленных зависимостей для конкретного проекта.

Заводить такое окружение лучше под каждый новый проект.

Такую проблему нам поможет решить **Git**. Он позволяет следить за версиями проекта, сохранять работающие и потом удобно возвращаться к ним, если все сломалось.

## Терминал

Начнем мы с терминала, так как для остальных двух тем нам уже потребуется его понимание.

### Что это вообще такое?

Три понятия, которые обычно смешиваются:

| Слово | Что это |
|-------|---------|
| **Терминал** | окно/программа, которая рисует текст и принимает клавиши. Iterm2, Windows Terminal, встроенный терминал VS Code |
| **Шелл (shell)** | программа *внутри* терминала, которая понимает команды: `bash`, `zsh`, `PowerShell`, `fish` |
| **CLI** | command-line interface — способ общения с программой текстом, а не кнопками |

Набор горячих клавиш, всякие функции, связанные с открытием новых окон и интеграцией с ОС (например, возможность нажать «открыть в терминале» в проводнике/finder) – это часть, за которую отвечает терминал.

Набор доступных команд и их синтаксис – это часть, за которую отвечает шелл. Причем в одном терминале может быть несколько шеллов, и иногда, если у вас не работает какая-то документация из интернета, имеет смысл проверить, какой шелл у вас используется.

CLI – часть конкретной программы / команды в шелл. Например, если вы устанавливаете библиотеку через `conda`, у вас спросят, точно ли надо устанавливать. Такой формат взаимодействия программы с вами – это CLI.

### Почему не мышкой в интерфейсе?

Четыре основных причины:

1. **Воспроизводимость.** «Нажми правой кнопкой, потом `Свойства`, потом...» невозможно переслать, плюс интерфейс у другого человека может быть кастомизирован и отличаться от вашего. `pip install pprint` можно скопировать в чат.
2. **Автоматизация.** Последовательность команд можно записать в файл и запускать одной кнопкой. Последовательность кликов придется каждый раз делать заново.
3. **Серверы.** На удаленном сервере нет рабочего стола.
4. **Инструментов просто нет в GUI.** У `git`, `docker`, `pip`, `pytest` интерфейса либо нет (как у `pip`), либо он сильно ограничен (как у `git`).

### Как устроена команда?

```
$ ls -la /Users/anna/projects
│  │  │   └── аргумент (над чем работаем)
│  │  └────── флаги/опции (как работаем)
│  └───────── команда (что делаем)
└──────────── приглашение (prompt), его выводит шелл, не вы
```

Про флаги:
* короткие: `-l`, `-a`. Они склеиваются: `-la` это то же самое, что `-l -a`
* длинные: `--help`, `--version`, `--global`
* значение: `-m "текст"` или `--message "текст"` (бывает еще `--message="текст"`, но редко и не везде поддерживается)

### Файловая система — это дерево

```
/                          ← корень (в Windows: C:\)
├── Users/
│   └── my_user/              ← ваш домашний каталог, он же ~
│       ├── Desktop/
│       └── projects/
│           └── my_app/
└── etc/
```

**У шелла всегда есть «текущая папка» (working directory).** Все относительные пути считаются от неё. Команда `pwd` отвечает на вопрос «где я?» и выводит абсолютный путь до текущей папки. Советуем использовать ее почаще, особенно если вдруг что-то работает не так, как вы ожидали.

На всякий случай:
- Относительный путь – путь, который считается от текущей директории. Чаще всего начинается сразу с названия папки. Например `projects/my_app`.
- Абсолютный путь – путь, который считается от корня файловой системы. Например `/Users/my_user/projects/my_app`.

Особые обозначения в путях:

| Обозначение | Смысл |
|--------|---------|
| `.` | текущая папка |
| `..` | папка на уровень выше |
| `../..` | на два уровня выше |
| `~` | домашний каталог (`/Users/my_user`) |
| `-` (в `cd -`) | предыдущая папка, где вы были |

Несколько советов:
- Когда работаете на общем устройстве (сервер команды или даже компьютер с двумя пользователями дома), старайтесь использовать пути относительно `~`, чтобы случайно не задеть чужие данные;
- Пути вида `../../../my_dir/file.txt` очень тяжело читаются глазами. Пожалуйста, либо избегайте их использования, либо комментируйте, что вы имели в виду.

### __⚠️ Будьте осторожны с `rm`__

```bash
rm -rf folder
```

* **Корзины нет.** Удалённое не восстанавливается.
* Подтверждения нет: `-f` = force, «не спрашивай».
* Один лишний пробел меняет смысл катастрофически: `rm -rf / home/my_user` вместо `rm -rf /home/my_user` удалит вам весь корневой каталог (а там ваша операционная система может лежать!).

**Заведите привычку перед `rm -rf` сначала выполнить `ls` с тем же путём. Убедились, что удаляете именно это — только потом `rm`.**

### Важно про эту тетрадку

Jupyter — это **не терминал**, хотя и позволяет запускать команды в нем. Но есть несколько нюансов:

| В терминале | В тетрадке |
|-------------|-----------|
| `ls -la` | `!ls -la` |
| `cd my_project` | `%cd my_project` ← **не** `!cd`, объяснение ниже |
| `source .venv/bin/activate` | **не работает**, только в терминале (это команда, которая активирует виртуальное окружение) |

**Почему `!cd` не работает?**

Каждая `!`-команда запускается в *отдельном* дочернем процессе. Он переходит в папку, печатает результат и умирает — а Jupyter остаётся там, где был. `%cd` — это «магия» Jupyter, она меняет папку самого ядра.

## Окружение и зависимости проекта

### Папка проекта

Проект живёт в **одной** папке. Не «в загрузках», не на рабочем столе вперемешку с мемами и скриншотами.

```bash
mkdir -p ~/projects/weather_app
cd ~/projects/weather_app
```

Типовая структура (к ней придём не сразу, но пусть будет ориентир):

```
weather_app/
├── .git/               ← служебная папка git (создастся сама), там он хранит информацию о версиях
├── .venv/              ← виртуальное окружение, иногда может быть снаружи
├── .gitignore          ← что git должен игнорировать (большие файлы, файлы с паролями и т.д.)
├── README.md           ← что это за проект и как запустить
├── requirements.txt    ← список зависимостей
├── src/                ← код
│   └── weather/
│       ├── __init__.py
│       └── main.py
├── tests/              ← тесты
└── data/               ← данные
```

Правила имён папок и файлов, которые избавят от многих проблем:
* только латиница, нижний регистр, `_` вместо пробелов: `weather_app`, не `Погода App`
* пробел в пути = вечная необходимость кавычек и экранирования
* кириллица в путях = сюрпризы с кодировками на другой ОС

### Виртуальное окружение

**Виртуальное окружение** — отдельная папка с личной копией Python и личной `site-packages` для одного проекта.

```
Системный Python  ──┬── .venv проекта A  
                    └── .venv проекта B
```

Правило: **один проект = одно окружение**.

__Как это работает?__

```bash
python -m venv .venv          # создать (иногда python3 -m venv .venv)
```
Читаем команду: «запусти модуль `venv` и создай окружение в папке `.venv`».

Активация — **зависит от ОС**:

| ОС / шелл | Команда |
|-----------|---------|
| macOS / Linux (bash, zsh) | `source .venv/bin/activate` |
| Windows PowerShell | `.venv\Scripts\Activate.ps1` |
| Windows cmd | `.venv\Scripts\activate.bat` |
| Windows Git Bash | `source .venv/Scripts/activate` |

В PowerShell может выдавать ошибку, связанную с правами доступа, можно выполнить это в cmd.

После активации приглашение меняется:
```
(.venv) my_user@mac weather_app %
 ^^^^^^ вот этот префикс = окружение активно, теперь все команды будут выполняться в нем
```

Проверка, что вы действительно внутри:
```bash
which python      # должен указывать в .../weather_app/.venv/bin/python
which pip
```

Выйти:
```bash
deactivate
```

Важно про виртуальные окружения

1. **`.venv` не коммитят в git.** Это сотни мегабайт бинарников, привязанных к вашей ОС, так что никому другому полезны они не будут. Даже вам после минимальных изменений в системе могут оказаться бесполезны. Так что в git кладут `requirements.txt`, а окружение каждый пересобирает у себя.
2. **Активация живёт в одной сессии терминала.** При открытии нового окна или перезапуске придется активировать заново.
3. **Если `(.venv)` в приглашении есть, а IDE запускает другой интерпретатор.** В VS Code: `Ctrl+Shift+P` → *Python: Select Interpreter* → выбрать `.venv`. По умолчанию там либо ничего, либо системный/предыдущий выбранный интерпретатор.
4. **Окружение нельзя переносить и переименовывать** — внутри абсолютные пути. Проще и правильнее удалить и создать заново.

### Зависимости

**Зависимость** — чужой код, на который опирается ваш. В привычном нам мире питона это библиотеки (`requests`, `pandas`, `flask`), хотя в широком смысле тот же питон тоже можно считать одной из зависимостей.

```bash
pip install requests                # последняя версия
pip install "requests==2.31.0"      # ровно эта версия
pip install "requests>=2.28,<3.0"   # любая версия из диапазона, чаще всего последняя из подходящих
pip list                            # что стоит в текущем окружении
pip show requests                   # подробности про библиотеку + от чего зависит сам пакет
pip uninstall requests
```

__Фиксация: `requirements.txt`__

```bash
pip freeze > requirements.txt       # выгрузить всё установленное с версиями
pip install -r requirements.txt     # поставить всё из файла (у себя/на сервере)
```

Файл — обычный текст:
```
requests==2.31.0
pandas==2.2.0
python-dotenv==1.0.1
```

**Этот файл обязательно класть в git.** Он и есть часть ответа на вопрос «а как это запустить другому человеку».

Ещё есть специальные библиотеки, чтобы собирать такие файлы. Например, `pipreqs`.

Проблема `pip freeze` в том, что он выгружает **всё** — и ваши библиотеки, и все случайно установленное, и все системные пакеты, знание о половине из которых не нужно никому, кроме разработчиков самого питона.

**`pipreqs`** решает это иначе: он сканирует исходный код и собирает только те пакеты, которые реально импортированы в ваших `.py`-файлах.

```bash
pip install pipreqs
pipreqs .                     # сканирует текущую папку → пишет requirements.txt
pipreqs . --force             # перезаписать существующий файл
pipreqs . --ignore .venv      # игнорировать папки
```

**Когда что использовать:**

| Инструмент | Принцип | Плюсы | Минусы |
|---|---|---|---|
| `pip freeze` | «что стоит в окружении» | 100% воспроизводимо, включая все версии. Полезно скорее для отладки и проверки, что у вас стоит | Шумно, очень много лишней информации |
| `pipreqs` | «что импортировано в коде» | Только нужное, выпишет библиотеки, даже если они у вас не установлены | Может пропустить runtime-зависимости, не определяет версии точно |
| более сложные библиотеки (например, `uv`) | Все сразу: и установка зависимостей, и их запись в файл со списком, и отслеживание их версий | Всё в одном | Нужно перейти на uv полностью, не совместим с прочими инструментами |

### Про версии

`2.31.4` = `MAJOR.MINOR.PATCH`
* `PATCH` — багфиксы, для пользователя ничего не меняется, кроме починенного бага
* `MINOR` — новые фичи, обратно совместимо (то есть ваш старый код будет работать, но еще будет что-то новое)
* `MAJOR` — большие изменения часто без обратной совместимости

| Запись | Смысл |
|--------|-------|
| `requests` | Любая версия. Удобно сегодня, больно через полгода |
| `requests==2.31.0` | Ровно эта. Максимальная предсказуемость, но у кого-то такая версия может просто не подходить под систему |
| `requests>=2.28` | Не старее. Часто, если используется какая-то функция, которая только в этой версии появилась |
| `requests~=2.31.0` | `>=2.31.0, <2.32.0` — только патчи, без больших изменений |

Практика: для приложений и проектов — жёсткий `==` (воспроизводимость важнее свежести). Для библиотек — диапазоны (иначе поломаете чужие сборки).

## Git

### Что это и зачем?

**Git — система контроля версий.**

Она сохраняет снимки состояния проекта и позволяет:

* вернуться к любому прошлому состоянию («вчера работало» — и правда вернуться во вчера);
* видеть, **кто, когда и зачем** изменил каждую строку;
* работать над одной кодовой базой втроём и не перетереть друг друга;
* вести эксперимент в **ветке**, не ломая рабочую версию;
* иметь копию всей истории на сервере.

__Git ≠ GitHub__

| | |
|---|---|
| **Git** | программа на вашем компьютере. Работает без интернета. Автор — Линус Торвальдс, 2005 |
| **GitHub / GitLab / Bitbucket** | сайты, где хранят копию репозитория. Плюс задачи, ревью, CI |

Git — язык. GitHub — один из мессенджеров, где на нём говорят.

### Как git смотрит на ваши файлы?

Git не сохраняет каждое ваше нажатие Ctrl+S. Он сохраняет только то, что вы **сами попросили сохранить**. Поэтому у файла есть три состояния:

```
   вы редактируете          вы отобрали             git записал
   файлы в папке      →     что сохранять     →     версию проекта

                    git add                git commit
```

| Состояние | Что это значит |
|---|---|
| **Изменён** | вы поправили файл, git это видит, но пока никуда не записал |
| **Добавлен** (`git add`) | вы сказали «вот это пойдёт в следующее сохранение» |
| **Закоммичен** (`git commit`) | версия записана в историю, её уже можно найти и вернуть |

__Зачем промежуточный шаг `git add`? Почему нельзя просто «сохранить всё»?__

Потому что за час работы вы обычно правите несколько разных вещей сразу: чуть-чуть поправили текст, заодно исправили опечатку в README, заодно поменяли настройку.

`git add` позволяет разложить это по отдельным сохранениям: отдельно «поправил текст», отдельно «исправил опечатку». Тогда через месяц вы сможете отменить одно, не тронув другое.

На практике в учебных проектах вы чаще всего будете писать `git add .` — «возьми всё, что я наменял». Это нормально. Но знать, что можно выбирать, полезно.

### Коммит

**Коммит — это сохранённая версия всего проекта** плюс подпись: кто, когда и что сделал.

```
коммит 1 ──── коммит 2 ──── коммит 3 ──── коммит 4
"начал"      "добавил      "поправил     "добавил
              загрузку"     опечатку"     графики"
```

У каждого коммита есть:
* **номер** (выглядит как `a3f5b21`) — по нему можно вернуться именно к этой версии;
* **автор и дата** — берутся из настроек, которые мы сейчас пропишем;
* **сообщение** — то, что вы напишете сами.

Аналогия: коммит — это сохранение в компьютерной игре. Вы играете дальше, но всегда можете загрузиться с любого сохранения.

Важное отличие от Ctrl+Z: отменять можно не «последние 20 действий», а **вернуться к конкретной версии недельной давности**, даже если после неё вы работали ещё три дня.

### Настройка (один раз на компьютер)

Git подписывает каждый ваш коммит, поэтому ему нужно знать, кто вы:

```bash
git config --global user.name "YOUR NAME"
git config --global user.email "pochta@example.com"
git config --global init.defaultBranch main
git config --list                      # проверить, что записалось
```

Почту лучше указывать ту же, что на GitHub — тогда коммиты свяжутся с вашим профилем.

`init.defaultBranch main` — чтобы основная ветка называлась `main`. Исторически она называлась `master`, вы встретите оба названия, но сейчас все переезжают на `main`.

! Имя и почта попадают в каждый коммит **навсегда**. Если сейчас не настроить, git либо будет ругаться, либо подпишет коммиты чем-то вроде `user@MacBook-Air.local`.

Также она выведет вам конкретную ветку (обсудим дальше) или коммит, на котором вы находитесь. Это полезно, если вы давно не заходили в этот проект и уже чуть-чуть забыли, над чем работали.

### .gitignore

Есть файлы, которым в репозитории не место. Чтобы git их не трогал, их перечисляют в файле `.gitignore`.

| Что | Почему не нужно хранить |
|-----|--------|
| `.venv/` | сотни мегабайт, привязано к вашей ОС, у другого человека всё равно не заработает |

### Сообщения коммитов

Сообщение пишется для человека — чаще всего для вас же через три месяца, когда вы будете искать, где что сломалось.

**Плохо:** `fix`, `.`, `работа`, `asdf`, `изменения`, `финальная версия 2`

**Хорошо:** `Add city search`, `Fix crash on empty response`, `Update README with setup steps`

Практические правила:
* пишите, **что изменилось**, а не «как» — «как» и так видно в самих изменениях;
* одно сохранение = одно логическое изменение;
* английский не обязателен, но принят почти везде; главное — единообразие внутри проекта.

Простой тест: если по вашему списку коммитов можно пересказать, что происходило с проектом — вы всё делаете правильно.

Есть распространённое соглашение **Conventional Commits**, где перед описанием ставят тип изменения:

```
feat: добавил поиск по городам
fix: починил падение на пустом ответе
docs: обновил инструкцию запуска
```

Знать его полезно, потому что вы встретите такой формат в чужих проектах. Требовать от себя прямо сейчас — не обязательно. В больших компаниях, где используются трекеры задач, в названиях коммитов часто принято послать еще и номер задачи, чтобы все источники информации внутри компании были связаны.

### Ветки

Пока вы работаете в одиночку и линейно, веток достаточно одной. Ветки нужны, когда появляется вопрос: **«я хочу попробовать, но боюсь сломать то, что работает»**. Или когда вас в команде несколько, и вы работаете над параллельными задачами.

Ветка — это отдельная линия работы. Вы отходите в сторону от основной версии, спокойно там всё меняете, а основная версия остаётся нетронутой.

```
                  добавил поиск ── поправил поиск     ← ветка feature/search
                 /
  A ── B ── C ───                                     ← main, рабочая версия
```

Пока вы в своей ветке:
* `main` продолжает работать, его можно кому-то показать в любой момент;
* если эксперимент не удался — просто удаляете ветку, и как будто ничего не было;
* если удался — **вливаете** ветку обратно в `main` (это называется merge).

Именно так работают команды: у каждого своя ветка под свою задачу, никто никому не мешает.

__Команды__

```bash
git branch                        # список веток, звёздочка = текущая
git switch -c feature/search      # создать ветку и сразу перейти в неё
git switch main                   # перейти в существующую ветку
git switch -                      # вернуться в предыдущую

# checkout — старый вариант, вы встретите его везде в интернете:
git checkout -b feature/search    # = git switch -c feature/search
git checkout main                 # = git switch main
```

`checkout` делает сразу две совершенно разные вещи в зависимости от аргумента: переключает ветку **или** восстанавливает файл. Это приводило к путанице, поэтому в git 2.23 разбили на две команды:
- `git switch` — переключение между ветками
- `git restore` — восстановление файлов

Какую из них использовать — без разницы, они делают одно и то же. Но если вы учитесь сейчас, проще сразу привыкнуть к `switch` и `restore`: у них понятнее назначение.

__Как называть ветки__

```
feature/user-login     новая функциональность
fix/empty-response     исправление ошибки
docs/setup-guide       правки в документации
```

Латиница, нижний регистр, дефисы вместо пробелов. Косая черта — просто для группировки, никакой магии в ней нет.

__Слияние__

```bash
git switch main                 # переходим туда, КУДА вливаем
git merge feature/search        # вливаем ветку
git branch -d feature/search    # удаляем ветку, она больше не нужна
```

Логика простая: сначала встаём в ветку-получатель, потом говорим, что в неё влить. Удалять влитую ветку не жалко — все коммиты уже в `main`.

### Удалённый репозиторий

До сих пор всё жило только на вашем компьютере. Чтобы появилась копия на GitHub (и возможность работать вместе), нужен **удалённый репозиторий**.

```bash
git remote add origin git@github.com:username/my_project.git   # привязать репозиторий на GitHub
git remote -v                      # проверить, к чему привязаны
git push -u origin main            # отправить: первый раз с -u
git push                           # отправить: все последующие разы
git pull                           # забрать изменения, которые сделали другие
git clone <ссылка>                 # скачать себе чужой или свой проект целиком
```

* `origin` — просто общепринятое название основного удалённого репозитория. Это не ключевое слово, можно назвать иначе, но не нужно.
* `-u` в первом push связывает вашу локальную ветку с веткой на сервере. После этого хватает короткого `git push`.
* (в терминалах Windows могут быть проблемы из-за слэша не в том направлении)

__push и pull__

* `push` — «отправить мои коммиты на сервер»
* `pull` — «забрать с сервера то, что сделали другие»

Коммит **не** попадает на GitHub автоматически. Пока вы не сделали `push`, ваша работа есть только у вас на ноутбуке. Это частая причина «я же всё сохранил, а препод ничего не видит».

__HTTPS или SSH__

Есть два способа подключаться к GitHub:

| | HTTPS | SSH |
|---|-------|-----|
| Ссылка | `https://github.com/user/repo.git` | `git@github.com:user/repo.git` |
| Как подтверждает, что это вы | логин + токен | ключ, настраивается один раз |
| В работе | периодически просит данные | больше ничего не спрашивает |

Обычный пароль от GitHub при работе с git не подойдёт — с 2021 года нужен либо специальный токен (Personal Access Token), либо SSH-ключ.

### Стандартный рабочий процесс

Вот то, ради чего был весь блок про git. Эту последовательность вас попросят выполнить на стажировке, на курсовой и в любом командном проекте.

```
1. git switch main            встаём в основную ветку
2. git pull                   забираем то, что сделали другие
3. git switch -c feature/x    заводим свою ветку под задачу
4. ...работаем...
5. git add .                  отбираем изменения
6. git commit -m "..."        сохраняем (шаги 4-6 повторяются много раз)
7. git push -u origin feature/x   отправляем ветку на GitHub
8. открываем Pull Request     просим влить нашу работу в main
9. коллеги смотрят, вы правите
10. ветку вливают в main
11. git switch main && git pull   обновляемся
12. git branch -d feature/x       убираем ненужную ветку
```

__Что такое Pull Request__

Pull Request (на GitHub) или Merge Request (на GitLab) — это **просьба влить вашу ветку в основную**. Создаётся кнопкой на сайте, не в терминале.

Зачем он нужен, если есть `git merge`:
* коллеги видят, что именно вы поменяли, и могут оставить комментарии до того, как изменения попадут в основную версию;
* автоматически прогоняются проверки;
* остаётся запись обсуждения: почему сделали именно так.

В одиночном учебном проекте можно обойтись `git merge` локально. В команде почти всегда работают через Pull Request.

Шаги 1–2 (перейти в main и забрать чужое) кажутся необязательными, пока вы один. В команде их пропуск — самая частая причина конфликтов: вы начинаете задачу от версии недельной давности.

### Шпаргалка: новый проект с нуля

```bash

mkdir -p ~/projects/my_app && cd ~/projects/my_app

# 2. ОКРУЖЕНИЕ
python -m venv .venv
source .venv/bin/activate            # Windows: .venv\Scripts\activate

# 3. ЗАВИСИМОСТИ
python -m pip install requests
python -m pip freeze > requirements.txt

# 4. ИГНОР-ЛИСТ (до git init)
printf '.venv/\n__pycache__/\n*.pyc\n.env\n.DS_Store\n' > .gitignore

# 5. РЕПОЗИТОРИЙ
git init
git add .
git commit -m "Initial commit"

# 6. ВЕТКА ПОД ЗАДАЧУ
git switch -c feature/first-feature
# ...работаем...
git add .
git commit -m "Add first feature"

# 7. ОТПРАВИТЬ НА GITHUB
git remote add origin git@github.com:username/my_app.git
git push -u origin main

# 8. ВЛИТЬ И ПОЧИСТИТЬ
git switch main
git merge feature/first-feature
git branch -d feature/first-feature
```
