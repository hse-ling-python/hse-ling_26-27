## Модуль `re`

Документация модуля – [здесь](https://docs.python.org/3/library/re.html). Модуль встроенный, ничего дополнительно устанавливать не надо.

```python
import re
```

**Какие функции поддерживает модуль `re`?** Разберем самые основные и полезные.

`findall(template, text)`

- возвращает список, состоящий из всех «мэтчей» (совпадений, найденных по шаблону)

```python
text = "The mouse was carefully disguised but caught quickly by the fox."

re.findall(r"\w+ly\b", text)  # Что выдаст код?
```

`search(template, text)`

- возвращает `Match object` – первое найденное совпадение, либо None (если ничего не было найдено)

- `Match object` – это объект, в котором хранится информация о поиске и о результате

    - у этого объекта есть атрибуты:
        -  `.span()` – возвращает пару, состояющую из начальной и конечной позиции мэтча
        -  `string` – возвращает строку, которая была передана в функцию
        - `.group()` – возвращает часть строки, в которой был найден мэтч

```python
text = "which foot or hand fell fastest"

my_match = re.search(r"\bf[a-z]*", text)  # Что выдаст код?
my_match
```

`finditer(template, text)`

- возвращает итератор по `Match object` – все найденные совпадения, по которым можно итерироваться
- атрибуты совпадают с предыдущим пунктом

```python
text = "which foot or hand fell fastest"

for match in re.finditer(r"\bf[a-z]*", text):
    print(match.group())
```

`split(template, text)`

- возвращает список из строк, которые получились в результате разделения строки по «мэтчу»

```python
text = (
    "Я люблю орехи!\n"
    "А какие орехи твои любимые?\n"
    "Например, мне нравится грецкие орехи. Или кешью"
)

re.split(r"[!?.]\s", text)  # Что выдаст код?
```

`sub(template, replace_string, text, count)`

- производит замену «мэтчей» на какую-то строку (можно регулировать количество замен)

```python
text = (
    "I had an uni e-mail which is vvtsareva_1@edu.hse.ru.\n"
    "Then I got an email as a teacher: vtsareva@hse.ru.\n"
    "Also there is a group email lingvo222@googlegroups.com! "
    "Now I don't check my e-mails at all..."
)

result = re.sub(
    r"[a-zA-Z0-9_.]+@[a-zA-Z0-9-]+(?:\.[a-zA-Z]{2,})+",
    "%email%",
    text,
)

print(result)  # Что выдаст код?
```
