### **Markdown**

Markdown – это упрощенный язык разметки, который преобразует (почти) обычный текст в HTML-страницу.

Создан в 2004 году Джоном Грубером и Аароном Шварцем, затем был дополнен и адаптирован для различных приложений.

Markdown-файлы имеют расширение `.md` или `.markdown`. Markdown используется на GitHub и, что особенно важно для нас сейчас – в Jupyter тетрадках для форматирования текстовых окон.

## Как и зачем работать с JSON?

### Что это?

**JSON** (*JavaScript Object Notation*) – простой, основанный на использовании текста, способ хранить и передавать структурированные данные.

- его придумали для того, чтобы упростить обмен данными
- его структура легко читается и составляется как человеком, так и компьютером
- его легко преобразовать в структуру данных для большинства языков программирования (числа, строки, логические переменные, массивы и т.д.)

Многие языки программирования имеют функции и библиотеки для чтения и создания структур JSON.

### Где это можно встретить?

**При работе с серверами**

Во втором модуле мы подробнее поговорим о том, как работает интернет, будем отправлять запросы и получать ответы с сайтов при помощи питона.

Как раз наиболее частое использование JSON – это пересылка данных от сервера к браузеру. Например, когда сервер отправляет браузеру веб-страницу, часто к странице прикладывается JSON с дополнительной информацией. Иногда весь ответ браузера состоит из JSON.

**При работе с библиотеками**

Существуют библиотеки для обработки естественного языка, которые выдают результат – например, морфологический разбор текста – в формате JSON. Чтобы вытащить из такого файла нужную информацию, нужно знать, как он устроен.

**В ваших проектах или домашних заданиях**

Очень часто оказывается, что именно в JSON удобно сложить все результаты, полученные в рамках домашнего задания / проекта / курсовой работы, для дальнейшего переиспользования (например, для визуализации) без перезапуска всего кода с нуля.


### Как устроен JSON?

Строка JSON может содержать __объект__, который начинается с `{` и заканчивается на `}`. Такой объект очень похож на питоновский *словарь*: у него есть ключи – строки, которые пишутся в кавычках, а через двоеточие пишется значение, пары ключ-значение разделяются запятыми.

**NB!** Если ключом в питоновском словаре может быть и число, и булевая переменная, и т.д. (не делайте так...), то при переводе подобного словаря в JSON-объект *все ключи* станут строками.

Например:

```json
{"first_name": "Guido", "last_name": "Rossum"}
```

Строка JSON может содержать __массив__, который начинается с `[` и заканчивается на `]`. Такой массив очень похож на питоновский список: в нем значения перечисляются через запятую.

Например:

```json
["Guido van Rossum", "Diana Clarke", "Naomi Ceder", "Van Lindberg", "Ewa Jodlowska"]
```



Значение в массиве или объекте может быть:
* числом (целым или с плавающей точкой)
* строкой (в двойных кавычках)
* логическим значением (`true` или `false`)
* другим массивом (заключенным в квадратные скобки)
* другим объектом (заключенным в фигурные скобки)
* значением null

Чтобы включить в строку специальные символы (например, кавычку), их нужно экранировать с помощью `\`, например, `\"` или `\r\n`. Наглядные правила построения JSON-строки можно посмотреть на [официальном сайте](http://www.json.org/), если захочется.

Может показаться, что это вообще-то все и так очень похоже на обычный питон. Но это не так.

Во-первых, JSON – это не исполняемый код, а просто текст.

Во-вторых, очень часто запись валидного питоновского словаря или массива не будет являться валидной записью в формате JSON.

Например, это не JSON, но при этом словарь: `{(1, 'a'): u'12345'}`.

**Сводная таблица конвертации данных Python в JSON**:

|    Python   |  JSON  |
|:-----------:|:------:|
| dict        | object |
| list, tuple | array  |
| str         | string |
| int, float  | number |
| True        | true   |
| False       | false  |
| None        | null   |

### Модуль `json`

#### Чтение

Для чтения в модуле `json` есть два метода:
- `json.load()` – метод считывает **файл** в формате JSON и возвращает объекты Python. У этой функции один обязательный аргумент – файл.
- `json.loads()` – метод считывает **строку** в формате JSON и возвращает объекты Python. У этой функции один обязательный аргумент – строка.

#### Запись

Для записи информации в формате JSON в модуле `json` также два метода:
- `json.dump()` – метод **записывает объект Python в файл** в формате JSON. У этой функции два обязательных аргумента – объект питона и файл.
- `json.dumps()` – **превращает** питоновский словарь или массив **в строку JSON**. У этой функции один обязательный аргумент – словарь или массив.


Под словом «файл» в данном случае имеется в виду любой файло-подобный объект – собственно файл, или любые объекты, к которым можно применить метод `.read()`.

#### Пример работы модуля

Попробуем превратить нашу строку в объект питона:

````python
json_string = """{"organisation": "Python Software Foundation",
                 "officers": [
                            {"first_name": "Guido", "last_name": "Rossum", "position":"president"},
                            {"first_name": "Diana", "last_name": "Clarke", "position":"chair"},
                            {"first_name": "Naomi", "last_name": "Ceder", "position": "vice chair"},
                            {"first_name": "Van", "last_name": "Lindberg", "position": "vice chair"},
                            {"first_name": "Ewa", "last_name": "Jodlowska", "position": "director of operations"}
                            ],
                "type": "non-profit",
                "country": "USA",
                "founded": 2001,
                "members": 244,
                "budget": 750000,
                "url": "www.python.org/psf/"}"""
````

````python
import json

data = json.loads(json_string)

# распечатаем тип объекта и убедимся, что теперь это не строка, а словарь
print(type(data))
````

````python
from pprint import pprint  # pretty-printer

# посмотрим на сам этот словарь
pprint(data)
````

````python
# и попробуем поработать с этим словарем. например, распечатаем его ключи.
for key in data:
    print(key, end=' ')
````

````python
# у нас есть питоновский словарь, который мы хотим сохранить в виде строки json
d = {"John": 51, "Kate": 12, "Bill": 27}
json_string = json.dumps(d)

# убедимся, что теперь наши данные превратились в строку
print(type(json_string))
````

````python
# распечатаем эту строку
print(json_string)
````

````python
# то же самое можно делать с массивами
arr = ['hello', 'world']
json_string = json.dumps(arr)
print(type(json_string))
print(json_string)
````

````python
# убедимся, что не все питоновские правильные объекты хорошо вписываются в json
d = {("A", 21): "John"}
json_string = json.dumps(d)
print(json_string)
````

#### Дополнительные параметры методов записи

1. `indent` – уровень отступа (может быть равен числу или строке). Он позволяет сделать так, чтобы данные записывались в файл с человекопонятным форматированием. Тогда файл можно будет открыть текстовым редактором и посмотреть глазами, что там внутри.

2. `ensure_ascii` – гарантия, что все не-ASCII символы (например, вся кириллица) в выводе будут экранированы. Дело в том, что если в ваших данных есть не-ASCII символы, то модуль JSON по умолчанию кодирует их специальным образом, используя при этом только символы из ограниченного набора, читающиеся одинаково почти во всех кодировках. Это хорошо при переносе данных из одной программы в другую: ничего не собьётся и не потеряется. Но это плохо для человека: понять, что в таком файле, станет невозможно.

3. `sort_keys` – сортировка ключей.

4. `skipkeys` – ключи словаря не базового типа будут проигнорированы. Это существует для случаев типа ячейкой выше, где ключом словаря быть кортеж, что вызвало `TypeError`.

````python
# вот такой код просто сбросит словарь в файл:
d = {'абв': 1, 'где': 2, 'ёжз': 3}

with open('data.json', 'w', encoding='utf-8') as f:
    json.dump(d, f)
````

Если заглянуть в файл, то результат будет таким:

```json
{"\u0433\u0434\u0435": 2, "\u0430\u0431\u0432": 1, "\u0451\u0436\u0437": 3}
```

````python
# добавим параметр ensure_ascii (у него дефолтное значение – True):
with open('data.json', 'w', encoding='utf-8') as f:
    json.dump(d, f, ensure_ascii=False)
````

Результат в файле:

```json
{"где": 2, "абв": 1, "ёжз": 3}
```

````python
# добавим indent (числовое значение - это число пробелов в отступах):
with open('data.json', 'w', encoding='utf-8') as f:
    json.dump(d, f, ensure_ascii=False, indent=4)
````

Результат в файле:

```json
{
    "абв": 1,
    "где": 2,
    "ёжз": 3
}
```

````python
# добавим кортеж в качестве одного из ключей и skipkeys:
d = {("A", 21): "John", 'абв': 1, 'где': 2, 'ёжз': 3}

with open('data.json', 'w', encoding='utf-8') as f:
    json.dump(d, f, ensure_ascii=False, indent=4, skipkeys=True)
````

Результат в файле все тот же, но ошибки `TypeError` больше нет:

```json
{
    "абв": 1,
    "где": 2,
    "ёжз": 3
}
```

### Пример получения json при работе с сервером

Если отправлять на GitHub специальные запросы по особым ссылкам, то в ответ сервер GitHub будет присылать JSON-строку с информацией. Например, можно посмотреть количество подписчиков или репозиториев у пользователя.

````python
import json
import urllib.request

user = "agricolamz"  # пользователь, про которого мы хотим что-то узнать
url = 'https://api.github.com/users/%s/repos' % user
# по этой ссылке мы будем доставать JSON, попробуйте вставить ссылку в браузер и посмотреть, что там

response = urllib.request.urlopen(url)  # посылаем серверу запрос и достаем ответ
text = response.read().decode('utf-8')  # читаем ответ в строку
data = json.loads(text) # превращаем json-строку в объекты питона

print(len(data))  # можно распечатать, сколько у пользователя репозиториев

for i in data:
    print(i["name"])  # и распечатать названия всех репозиториев
````
